```sql
%sql
CREATE CATALOG IF NOT EXISTS urban_mobility
COMMENT 'Real-Time Urban Mobility Intelligence Lakehouse';

USE CATALOG urban_mobility;

CREATE SCHEMA IF NOT EXISTS bronze
COMMENT 'Raw ingested data, close to source, immutable';

CREATE SCHEMA IF NOT EXISTS silver
COMMENT 'Cleaned, validated, deduplicated, enriched data';

CREATE SCHEMA IF NOT EXISTS gold
COMMENT 'Business-level aggregated data products';

CREATE SCHEMA IF NOT EXISTS quarantine
COMMENT 'Invalid records that failed data quality rules';

CREATE SCHEMA IF NOT EXISTS monitoring
COMMENT 'Pipeline runs, data quality results, streaming metrics';

SHOW SCHEMAS IN urban_mobility;
```

```sql
CREATE VOLUME IF NOT EXISTS urban_mobility.bronze.source_data
COMMENT 'Raw historical NYC TLC parquet files used as replay source for simulator';

CREATE VOLUME IF NOT EXISTS urban_mobility.bronze.landing
COMMENT 'Raw JSON event batches from simulator';

CREATE VOLUME IF NOT EXISTS urban_mobility.bronze.reference_data
COMMENT 'Generated/sourced reference data landing: drivers, vehicles, zones';

CREATE VOLUME IF NOT EXISTS urban_mobility.bronze.checkpoints
COMMENT 'Structured Streaming checkpoint locations';

CREATE VOLUME IF NOT EXISTS urban_mobility.bronze.schemas
COMMENT 'Auto Loader inferred schema storage';

SHOW VOLUMES IN urban_mobility.bronze;
```
