```python
from pyspark.sql import functions as F

df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("/Volumes/urban_mobility/bronze/reference_data/drivers/")
    .withColumn("_source_file", F.col("_metadata.file_path"))
    .withColumn("ingested_at", F.current_timestamp())
)

df.write.mode("overwrite").format("delta").saveAsTable("urban_mobility.bronze.drivers")

print("bronze.drivers rows:", spark.table("urban_mobility.bronze.drivers").count())
```
