```python
from pyspark.sql import functions as F

df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("/Volumes/urban_mobility/bronze/reference_data/vehicles/")
    .withColumn("_source_file", F.lit("/Volumes/urban_mobility/bronze/reference_data/vehicles/"))
    .withColumn("_ingested_at", F.current_timestamp())
)

df.write.mode("overwrite").format("delta").saveAsTable("urban_mobility.bronze.vehicles")

print("bronze.vehicles rows:", spark.table("urban_mobility.bronze.vehicles").count())
```
