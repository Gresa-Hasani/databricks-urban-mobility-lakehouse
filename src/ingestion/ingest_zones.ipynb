```python
import pandas as pd
from pyspark.sql import functions as F

zones_pd = pd.read_csv("https://d37ci6vzurychx.cloudfront.net/misc/taxi_zone_lookup.csv")
zones_pd = zones_pd.rename(columns={
    "LocationID": "zone_id",
    "Borough": "borough",
    "Zone": "zone_name",
    "service_zone": "service_zone"
})

df = (
    spark.createDataFrame(zones_pd)
    .withColumn("_source_file", F.lit("nyc_tlc_taxi_zone_lookup.csv"))
    .withColumn("_ingested_at", F.current_timestamp())
)

df.write.mode("overwrite").format("delta").saveAsTable("urban_mobility.bronze.zones")

print("bronze.zones rows:", spark.table("urban_mobility.bronze.zones").count())
```
