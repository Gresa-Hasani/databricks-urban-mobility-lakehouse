```python
from pyspark.sql import functions as F

# silver.drivers
drivers_silver = (
    spark.table("urban_mobility.bronze.drivers")
    .dropDuplicates(["driver_id"])
    .filter(F.col("driver_id").isNotNull())
    .withColumn("rating", F.round(F.col("rating"), 2))
)
drivers_silver.write.mode("overwrite").format("delta").saveAsTable("urban_mobility.silver.drivers")

# silver.vehicles
vehicles_silver = (
    spark.table("urban_mobility.bronze.vehicles")
    .dropDuplicates(["vehicle_id"])
    .filter(F.col("vehicle_id").isNotNull())
)
vehicles_silver.write.mode("overwrite").format("delta").saveAsTable("urban_mobility.silver.vehicles")

# silver.zones
zones_silver = (
    spark.table("urban_mobility.bronze.zones")
    .dropDuplicates(["zone_id"])
    .filter(F.col("zone_id").isNotNull())
)
zones_silver.write.mode("overwrite").format("delta").saveAsTable("urban_mobility.silver.zones")

print("silver.drivers:", spark.table("urban_mobility.silver.drivers").count())
print("silver.vehicles:", spark.table("urban_mobility.silver.vehicles").count())
print("silver.zones:", spark.table("urban_mobility.silver.zones").count())
```
