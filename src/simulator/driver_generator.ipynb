```python
import random
import string
import pandas as pd
from datetime import datetime, timedelta

random.seed(42)

zones_pd = pd.read_csv("https://d37ci6vzurychx.cloudfront.net/misc/taxi_zone_lookup.csv")

FIRST_NAMES = ["James","Mary","John","Patricia","Robert","Jennifer","Michael","Linda","William","Elizabeth","David","Barbara","Richard","Susan","Joseph","Jessica"]

LAST_NAMES = ["Smith","Johnson","Williams","Brown","Jones","Garcia","Miller","Davis","Rodriguez","Martinez","Hernandez","Lopez","Gonzalez","Wilson","Anderson"]

def random_phone():
    return f"+1{random.randint(200,999)}{random.randint(1000000,9999999)}"

N_DRIVERS = 10000
drivers = []
for i in range(1, N_DRIVERS + 1):
    fname, lname = random.choice(FIRST_NAMES), random.choice(LAST_NAMES)
    hire_date = datetime(2019,1,1) + timedelta(days=random.randint(0, 2500))
    drivers.append({
        "driver_id": f"DRV{i:06d}",
        "first_name": fname,
        "last_name": lname,
        "phone": random_phone(),
        "email": f"{fname.lower()}.{lname.lower()}{i}@mobilitydrivers.com",
        "phone_number": random_phone(),
        "license_number": f"LIC{random.randint(10000000,99999999)}",
        "hire_date": hire_date.strftime("%Y-%m-%d"),
        "rating": round(random.uniform(3.5, 5.0), 2),
        "home_zone_id": int(zones_pd["LocationID"].sample(1).values[0]),
        "status": random.choices(["ACTIVE","INACTIVE"], weights=[0.92,0.08])[0]
    })

drivers_df = spark.createDataFrame(pd.DataFrame(drivers))
print("Drivers generated:", drivers_df.count())

drivers_df.coalesce(1).write.mode("overwrite").option("header", True).csv("/Volumes/urban_mobility/bronze/reference_data/drivers/")

print("Done writing drivers landing file.")
```
