```python
import random
import string
import pandas as pd

random.seed(42)

MAKES_MODELS = [
    ("Toyota","Camry"), ("Toyota","Corolla"), ("Honda","Accord"), ("Honda","Civic"),
    ("Ford","Fusion"), ("Nissan","Altima"), ("Hyundai","Sonata"), ("Chevrolet","Malibu"),
    ("Kia","Optima"), ("Tesla","Model 3")
]

def random_plate():
    return "".join(random.choices(string.ascii_uppercase, k=3)) + str(random.randint(1000,9999))

N_DRIVERS = 10000
N_VEHICLES = 10000

# 1-per-1 assignment: shuffle driver_ids, zip with vehicles -> full coverage guaranteed
driver_ids = [f"DRV{i:06d}" for i in range(1, N_DRIVERS + 1)]
random.shuffle(driver_ids)

vehicles = []
for i in range(1, N_VEHICLES + 1):
    make, model = random.choice(MAKES_MODELS)
    vehicles.append({
        "vehicle_id": f"VEH{i:06d}",
        "make": make,
        "model": model,
        "year": random.randint(2016, 2025),
        "plate_number": random_plate(),
        "capacity": random.choice([4,4,4,6]),
        "vehicle_type": random.choices(["STANDARD","XL","PREMIUM"], weights=[0.75,0.15,0.10])[0],
        "assigned_driver_id": driver_ids[i - 1]
    })

vehicles_df = spark.createDataFrame(pd.DataFrame(vehicles))
print("Vehicles generated:", vehicles_df.count())
print("Distinct assigned drivers:", vehicles_df.select("assigned_driver_id").distinct().count())

vehicles_df.coalesce(1).write.mode("overwrite").option("header", True).csv(
    "/Volumes/urban_mobility/bronze/reference_data/vehicles/"
)
print("Done writing vehicles landing file.")
```
