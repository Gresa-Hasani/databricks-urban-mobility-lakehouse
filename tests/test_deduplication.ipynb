```python
import pytest
from pyspark.sql import SparkSession, Row

def test_duplicate_event_ids_removed(spark):
    df = spark.createDataFrame([
        Row(event_id="E1", trip_id="T1"),
        Row(event_id="E1", trip_id="T1"),
        Row(event_id="E2", trip_id="T2"),
    ])
    deduped = df.dropDuplicates(["event_id"])
    assert deduped.count() == 2


def test_valid_event_accepted_once(spark):
    df = spark.createDataFrame([
        Row(event_id="E3", trip_id="T3", fare_amount=25.0),
    ])
    deduped = df.dropDuplicates(["event_id"])
    assert deduped.count() == 1
    assert deduped.collect()[0]["fare_amount"] == 25.0
```

```python
def run_test(test_func, name):
    try:
        test_func(spark)
        print(f"✅ PASS: {name}")
    except AssertionError as e:
        print(f"❌ FAIL: {name} — {e}")
    except Exception as e:
        print(f"⚠️ ERROR: {name} — {e}")

run_test(test_duplicate_event_ids_removed, "test_duplicate_event_ids_removed")
run_test(test_valid_event_accepted_once, "test_valid_event_accepted_once")
```
